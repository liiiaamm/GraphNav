```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G=nx.Graph()
G.add_node(1)
G.add_node(2)
G.add_edge(1,2)
G.add_nodes_from([3,4,5])
G.add_edges_from([(1,4),(3,4),(2,4),(3,2),(3,5)])
print(G)
print(G.nodes)
print(G.edges)
```

```python
#Eliminamos nodo
G.remove_node(5)
print(G)
print(G.nodes)
print(G.edges)
```

```python
#Adyacencia
print(G[1])
print(G.adj[1])
```

```python
plot=plt.plot()
# nx.draw(G)
#nx.draw(G,with_labels=True)
nx.draw(G,with_labels=True,pos={1:(0,0),2:(0,1),3:(1,0),4:(1,1)})
plt.show()
```
